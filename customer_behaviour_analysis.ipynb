```python
import pandas as pd

df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# create a columns age_group

labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age', 'age_group']].head(10)
```

```python
# create columns purchase frequency dates

frequency_mapping = {
    'Fortnightly' : 14,
    'Weekly' : 7,
    'Quarterly' : 90,
    'Bi-Weekly' : 14,
    'Annually' : 365,
    'Every 3 Months' : 90
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
    
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied', 'promo_code_used']].head(10)
```

```python
(df['discount_applied'] == df ['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
df.columns
```

```python
from sqlalchemy import create_engine

# 1. Convert any categorical/interval columns to string format so SQL can accept them
df['age_group'] = df['age_group'].astype(str)

# 2. Create the SQLAlchemy engine (using pymysql driver)
engine = create_engine("mysql+pymysql://root:Param%40123@127.0.0.1:3306/customer_behaviour")

# 3. Transfer the processed DataFrame into MySQL
df.to_sql(
    name='customer',        # Table name in your database
    con=engine,             # SQLAlchemy engine connection
    index=False,            # Skips saving the pandas row index
    if_exists='replace'     # Replaces the table if it already exists
)

print("Data successfully transferred to MySQL!")
```
